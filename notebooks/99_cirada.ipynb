```python
from astropy import units as u
from astropy.coordinates import SkyCoord
from astroquery.cadc import Cadc

# Inicializa o cliente CADC
cadc = Cadc()

# 1. (Opcional) Listar as tabelas disponíveis para confirmar o nome correto
# As tabelas do CIRADA VLASS geralmente começam com vlass ou cirada
tables = cadc.get_tables()
vlass_tables = [t.name for t in tables if "vlass" in t.name.lower()]
print("Tabelas VLASS encontradas:", vlass_tables)

# 2. Definir a coordenada do seu alvo (Ex: 3C 273 ou coordenadas RA/Dec)
coords = SkyCoord(ra=187.277915 * u.deg, dec=2.052388 * u.deg, frame="icrs")

# 3. Realizar a consulta por região em um raio de 1 arcmin no catálogo do VLASS
# Substitua 'vlass.interim_component' pelo nome exato da tabela do CIRADA/VLASS desejada
result_table = cadc.query_region(coords, radius=1 * u.arcmin)

# Exibe os resultados
print(result_table)
```

```python
from astropy import units as u
from astropy.coordinates import SkyCoord
from astroquery.vizier import Vizier

# Configura o Vizier para não limitar o número de linhas (padrão é 50)
v = Vizier(row_limit=-1)

# Coordenadas do alvo
coords = SkyCoord(ra=187.277915, dec=2.052388, unit=(u.deg, u.deg))

# Query usando o identificador do catálogo CIRADA VLASS no VizieR (Ex: J/ApJS/255/30)
result = v.query_region(coords, radius=2 * u.arcmin, catalog="J/ApJS/255/30")

# O VizieR retorna uma lista de tabelas (ex: componentes, subtiles)
if result:
    result[0].pprint()
```
