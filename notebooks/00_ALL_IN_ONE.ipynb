## Radiotelescopio

- Mostra como inicializar radiotelescopio com settings
- Mostra como inicializar radiotelescopio com parametros
- inicializa radiotelescópio
- cria timerange
- skychart em coordenada equatorial mostrando apontamentos levando em conta FWHM

```python
"""Radiotelescópio: settings, parâmetros explícitos, TimeRange e skychart equatorial."""

from __future__ import annotations

import numpy as np
import matplotlib.pyplot as plt

# ── 1. Inicializa com settings.toml ──
from radioskysims.infrastructure.config.config_loader import (
    build_radiotelescope_from_settings,
)

rt = build_radiotelescope_from_settings("bingo")
print("== Radiotelescópio (settings) ==")
print("  nome:", rt.name)
print("  antenas:", [a.name for a in rt.antennas])
print("  observer:", rt.observer.lat.to_value("deg"), rt.observer.lon.to_value("deg"))

# ── 2. Inicializa com parâmetros explícitos ──
from radioskysims.domain.common import FrequencyRange, Observer, TimeRange
from radioskysims.domain.radiotelescope.antenna.models import (
    Antenna,
    BeamSpec,
    Pointing,
)
from radioskysims.domain.radiotelescope.backend.spectrometer.models import (
    SpectrometerBackend,
)
from radioskysims.domain.radiotelescope.specs import RadiotelescopeSpec

tr_expl = TimeRange.from_iso("2026-07-25T00:00:00", "2026-07-26T00:00:00", resolution=600)
fr_expl = FrequencyRange.from_floats(980.0, 1360.0, 1.0)
ant_expl = Antenna(
    name="BINGO Horn",
    frequency_range=fr_expl,
    pointing=Pointing(mode="transit", alt_deg=80.0, az_deg=0.0),
    beam=BeamSpec(
        name="LIT",
        polarizations=["H"],
        profile="gaussian",
        beam_width_deg=0.5,
    ),
    backend=SpectrometerBackend(
        name="Backend",
        frequency_range=fr_expl,
        num_channels=1024,
    ),
)
rt_expl = RadiotelescopeSpec(
    name="BINGO Explícito",
    observer=Observer(lat=-7.25, lon=-35.86, height=550.0, timeframe=tr_expl),
    antennas=[ant_expl],
)
print("\n== Radiotelescópio (parâmetros explícitos) ==")
print(rt_expl)

# ── 3. Cria TimeRange ──
tr = TimeRange.from_iso("2026-07-25T00:00:00", "2026-07-25T06:00:00", resolution=600)
print("\n== TimeRange ==")
print("  pontos:", len(tr.value))
print("  resolução:", tr.resolution, "s")

# ── 4. Skychart equatorial com pointing + FWHM ──
from radioskysims.domain.simulation.pointing import build_pointing_track
from radioskysims.infrastructure.beams.loaders.csv import CSVBeamLoader

antenna = rt.antennas[0]
obs = rt.observer
pt = build_pointing_track(antenna, tr_expl, obs)
beam_rt = CSVBeamLoader(file=antenna.beam.beam_pattern_file).load()
fwhm_deg = float(np.nanmean(np.asarray(beam_rt.params().fwhm, dtype=float)))

fig = plt.figure(figsize=(12, 6))
ax = fig.add_subplot(111, projection="mollweide")
ra_mw = ((pt["ra"].values + np.pi) % (2 * np.pi)) - np.pi
dec = pt["dec"].values
ax.scatter(ra_mw, dec, s=20, c="tab:red", label="Pointing (BINGO)")
th = np.linspace(0, 2 * np.pi, 100)
ax.plot(
    ra_mw[0] + np.deg2rad(fwhm_deg) * np.cos(th),
    dec[0] + np.deg2rad(fwhm_deg) * np.sin(th),
    "b-",
    alpha=0.6,
    label=f"FWHM ≈ {fwhm_deg:.2f}°",
)
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right")
ax.set_title("Skychart equatorial — pointing BINGO")
plt.tight_layout()
plt.show()
```

## Feixes

- imports 
- fix para root dir
- Cria feixe analítico gaussiano com FWHM de 30 - 45 arcmin para as frequencias de 980 - 1360 Mhz (bin de 1 MHz) (com dependencia linear com a frequencia)
- Mostra propriedades do objeto
- Visualiza beam pattern para todas as frequencias com holoviews. clique na curva mostra linha com FWHM correspondente no gráfico e ganho no tooltipb

```python
"""Feixes: cria feixe analítico gaussiano com FWHM 30–45 arcmin (980–1360 MHz)."""

from __future__ import annotations

import numpy as np
import xarray as xr

from radioskysims.domain.beam import Beam

# ── Fix do root (já feito no setup global, mantém robustez se rodado isolado) ──
# ── Feixe gaussiano com FWHM variando LINEARMENTE com a frequência ──
# FWHM: 45 arcmin (0.75°) @ 980 MHz  →  30 arcmin (0.5°) @ 1360 MHz
freqs_mhz = np.arange(980, 1361, 1)  # bin de 1 MHz
a_fwhm = (0.5 - 0.75) / (1360 - 980)  # deg/MHz
b_fwhm = 0.75 - a_fwhm * 980
fwhm_deg = a_fwhm * freqs_mhz + b_fwhm
sigma_deg = fwhm_deg / (2 * np.sqrt(2 * np.log(2)))

angles_deg = np.linspace(-3, 3, 601)
amp = np.exp(-0.5 * (angles_deg[:, None] / sigma_deg[None, :]) ** 2)
amp = amp / amp.max()

ds_beam = xr.Dataset(
    {"amplitude": (("angle", "frequency", "polarization"), amp[..., None])},
    coords={
        "angle": angles_deg,
        "frequency": freqs_mhz * 1e6,  # Hz
        "polarization": ["H"],
    },
)
beam = Beam(ds_beam)
print("== Feixe gaussiano (FWHM linear × f) ==")
print("  FWHM:", fwhm_deg.min(), "..", fwhm_deg.max(), "deg")
print("  freqs:", freqs_mhz.min(), "..", freqs_mhz.max(), "MHz")

# ── Mostra propriedades do objeto ──
params = beam.params()
print("  FWHM (deg) min/max:", np.nanmin(params.fwhm), "..", np.nanmax(params.fwhm))
print("  Directivity (dB):", params.directivity_db)
print("  Polarizações:", params.polarizations)
print("  Frequências (Hz):", params.frequencies[:5], "...")

# ── Visualiza beam pattern com holoviews (clique mostra linha FWHM + ganho tooltip) ──
import holoviews as hv
from holoviews import streams

curves = []
for f_mhz in [980, 1100, 1230, 1360]:
    f_hz = f_mhz * 1e6
    da = beam.dataset.sel(frequency=f_hz, polarization="H")["amplitude"]
    c = hv.Curve(
        (da.angle.values, da.values),
        kdims="angle (deg)",
        vdims="ganho",
        label=f"{f_mhz} MHz",
    ).opts(tools=["hover"])
    curves.append(c)

overlay = hv.Overlay(curves).opts(
    legend_position="top_right",
    width=900,
    height=400,
    title="Beam gaussiano — FWHM varia com a frequência (clique p/ FWHM)",
)

tap = streams.Tap(source=overlay, x=0.0, y=0.0)


def fwhm_marker(x, y):
    if x is None:
        return hv.VLine(0.0).opts(color="red", line_dash="dashed")
    # Frequência cujo FWHM (ângulo em que amplitude=0.5) está mais próxima do clique
    best_f = None
    best_d = np.inf
    for f_mhz in [980, 1100, 1230, 1360]:
        idx = int(np.argmin(np.abs(angles_deg - sigma_deg[freqs_mhz == f_mhz][0])))
        d = abs(angles_deg[idx] - x)
        if d < best_d:
            best_d = d
            best_f = f_mhz
    idx_fwhm = int(
        np.argmin(
            np.abs(
                angles_deg
                - sigma_deg[freqs_mhz == best_f][0] * np.sqrt(2 * np.log(2))
            )
        )
    )
    return hv.VLine(angles_deg[idx_fwhm]).opts(
        color="red", line_dash="dashed", line_width=2
    )


dmap = hv.DynamicMap(fwhm_marker, streams=[tap])
(overlay * dmap).opts(title="Clique numa curva → linha no FWHM correspondente")
```

- COnsulta feixes no repositorio
- carrega  "radioskysims/resources/beams/beam_example.csv"
- visualização como acima
- gráficos de FWHM x frequencia e Ganho por frequencia

```python
"""Feixes: consulta no repositório, carrega CSV e gráficos FWHM×f e Ganho×f."""

from __future__ import annotations

import numpy as np
import matplotlib.pyplot as plt

# ── Consulta feixes no repositório (PostgreSQL) ──
from radioskysims.infrastructure.db.beam_repository import BeamRepository
from radioskysims.infrastructure.db.engine import Database

db = Database()

# Garante que a tabela beamparamsdb existe (cria se necessário)
from sqlmodel import SQLModel

from radioskysims.infrastructure.db.beam_db_model import BeamParamsDB

SQLModel.metadata.create_all(db.engine)

with db.session() as session:
    repo = BeamRepository(session)
    beams = repo.list_all()
print(f"Feixes no banco (beamparamsdb): {len(beams)}")

# ── Carrega beam_example.csv ──
from radioskysims.infrastructure.beams.loaders.csv import CSVBeamLoader

beam_csv = CSVBeamLoader(file="radioskysims/resources/beams/beam_example.csv").load()
print("== beam_example.csv ==")
print("  frequências (Hz):", beam_csv.dataset.frequency.values)
print("  polarizações:", beam_csv.dataset.polarization.values)

params_csv = beam_csv.params()
freqs_csv_hz = np.asarray(params_csv.frequencies, dtype=float)
freqs_csv_mhz = freqs_csv_hz / 1e6
print("  FWHM (deg):", params_csv.fwhm)
print("  Directivity (dB):", params_csv.directivity_db)

# ── Visualização como acima (holoviews) ──
import holoviews as hv

curves_csv = []
for pol in beam_csv.dataset.polarization.values:
    for f_hz in beam_csv.dataset.frequency.values:
        da = beam_csv.dataset.sel(polarization=pol, frequency=f_hz)["amplitude"]
        curves_csv.append(
            hv.Curve(
                (da.angle.values, da.values),
                kdims="angle (deg)",
                vdims="amplitude",
                label=f"{pol} {f_hz/1e6:.0f} MHz",
            ).opts(tools=["hover"])
        )

hv.Overlay(curves_csv).opts(
    legend_position="top_right",
    width=900,
    height=400,
    title="Beam CSV (beam_example.csv)",
)

# ── Gráficos FWHM × frequência e Ganho × frequência ──
fwhm_csv = np.asarray(params_csv.fwhm, dtype=float)
fwhm_flat = np.nanmean(fwhm_csv, axis=-1)  # média sobre polarizações
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(freqs_csv_mhz, fwhm_flat, "o-")
axes[0].set_xlabel("Frequência (MHz)")
axes[0].set_ylabel("FWHM (deg)")
axes[0].set_title("FWHM × frequência")
axes[0].grid(True, alpha=0.3)

axes[1].plot(freqs_csv_mhz, np.full_like(freqs_csv_mhz, params_csv.directivity_db), "s-")
axes[1].set_xlabel("Frequência (MHz)")
axes[1].set_ylabel("Ganho (dB)")
axes[1].set_title("Ganho × frequência")
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Radiosources

- Demonstra uso do repositorio
- Faz query de fontes com fluxo maior do que 10Jy
- Skychart Mollweide em coordenadas equatoriais com radiofontes
- Grid com fluxo x frequencia para as 4 fontes mais brilhantes (2 x 2)

```python
"""Radiosources: repositório, query > 10 Jy, skychart Mollweide e grid espectral."""

from __future__ import annotations

import numpy as np
import matplotlib.pyplot as plt

from radioskysims.application.services.catalog_services import (
    RadioSourceCatalogService,
)
from radioskysims.domain.radiosources import (
    RadioSource,
    RadioSourceSignalEvaluator,
    SpectralModel,
)
from radioskysims.infrastructure.db.engine import Database
from radioskysims.infrastructure.db.radiosources_repository import (
    PostgresRadioSourcesRepository,
)
from radioskysims.infrastructure.external.radiosources_ingest import (
    VizierTapIngestAdapter,
)

db = Database()
pg_repo = PostgresRadioSourcesRepository(session_factory=db.session)
catalog = RadioSourceCatalogService(
    ingest_port=VizierTapIngestAdapter(
        tap_url="https://tapvizier.u-strasbg.fr/TAPVizieR/tap"
    ),
    repository=pg_repo,
)
print("== Repositório de RadioSources (PostgreSQL) ==")

# ── Query de fontes na região ──
sources = catalog.list_sources_in_region(ra_deg=190.0, dec_deg=0.0, radius_deg=90.0)
print(f"Fontes na região (RA=190°, Dec=0°, r=90°): {len(sources)}")

freqs_mhz = np.linspace(980, 1360, 50)
flux = catalog.evaluate_flux(sources, freqs_mhz)
strong = sorted(sources, key=lambda s: float(np.mean(flux[s.id])), reverse=True)
strong = [s for s in strong if float(np.mean(flux[s.id])) > 10.0]
print(f"Fontes com fluxo médio > 10 Jy no banco: {len(strong)}")

# Fallback: fontes canônicas fortes (CygA, CasA, VirA, HerA) se o banco não tiver
if not strong:
    canonical = [
        RadioSource(
            id="CygA",
            ra_deg=299.868,
            dec_deg=40.734,
            spectral_model=SpectralModel(nu0_mhz=1400.0, s0_jy=1570.0, alpha=0.75),
        ),
        RadioSource(
            id="CasA",
            ra_deg=350.867,
            dec_deg=58.815,
            spectral_model=SpectralModel(nu0_mhz=1400.0, s0_jy=2200.0, alpha=0.77),
        ),
        RadioSource(
            id="VirA",
            ra_deg=187.706,
            dec_deg=12.391,
            spectral_model=SpectralModel(nu0_mhz=1400.0, s0_jy=370.0, alpha=0.83),
        ),
        RadioSource(
            id="HerA",
            ra_deg=250.42,
            dec_deg=31.676,
            spectral_model=SpectralModel(nu0_mhz=1400.0, s0_jy=390.0, alpha=1.05),
        ),
    ]
    strong = canonical
    print("Usando fontes canônicas fortes para a demonstração.")

display_sources = strong[:6]
flux_mean = {
    s.id: float(np.mean(RadioSourceSignalEvaluator.evaluate_flux(s, freqs_mhz)))
    for s in display_sources
}
print("Fontes > 10 Jy:")
for s in display_sources:
    print(f"  {s.id:8s} RA={s.ra_deg:7.3f}°  Dec={s.dec_deg:6.3f}°  <S>={flux_mean[s.id]:9.1f} Jy")

# ── Skychart Mollweide em coordenadas equatoriais ──
fig = plt.figure(figsize=(12, 6))
ax = fig.add_subplot(111, projection="mollweide")
ra_rad = np.deg2rad([s.ra_deg for s in display_sources])
dec_rad = np.deg2rad([s.dec_deg for s in display_sources])
sc = ax.scatter(
    ((ra_rad + np.pi) % (2 * np.pi)) - np.pi,
    dec_rad,
    s=50,
    c=[flux_mean[s.id] for s in display_sources],
    cmap="inferno",
    edgecolors="k",
)
ax.grid(True, alpha=0.3)
ax.set_title("Skychart Mollweide — radiofontes (>10 Jy)")
plt.colorbar(sc, label="Fluxo (Jy)")
plt.tight_layout()
plt.show()

# ── Grid fluxo × frequência (4 fontes mais brilhantes, 2×2) ──
top4 = display_sources[:4]
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, s in zip(axes.flat, top4):
    f = RadioSourceSignalEvaluator.evaluate_flux(s, freqs_mhz)
    ax.plot(freqs_mhz, f, linewidth=2)
    ax.set_title(f"{s.id}  (S₀={s.spectral_model.s0_jy:.0f} Jy, α={s.spectral_model.alpha:.2f})")
    ax.set_xlabel("Frequência (MHz)")
    ax.set_ylabel("Fluxo (Jy)")
    ax.grid(True, alpha=0.3)
plt.suptitle("Fluxo × frequência — 4 fontes mais brilhantes")
plt.tight_layout()
plt.show()
```

## LocalSource

- Define timerange
- Define observador
- Uso do repositorio
- Obtem fluxo solar no intervalo
- Grafico de fluxo solar freq x tempo com intensidade em colorbar
- Refaz o grafico skychart mollweide com as radiofontes e o sol, colocando as posições ocupadas pelo sol no intervalo de tempo. Usar valores sensatos para redesenhar o sol, primeiro e último bem marcados e os outros intermediários com alpha 0.2

```python
"""LocalSource: fluxo solar no intervalo + skychart com radiofontes e o Sol."""

from __future__ import annotations

from datetime import UTC, datetime

import matplotlib.pyplot as plt
import numpy as np

from radioskysims.application.services.local_sources_service import (
    LocalSourcesService,
)
from radioskysims.domain.common import Observer, TimeRange
from radioskysims.infrastructure.db.solar_flux_repository import SolarFluxRepository
from radioskysims.infrastructure.external.local_source_track_skyfield_provider import (
    SkyfieldLocalSourceTrackProvider,
)
from radioskysims.infrastructure.external.skyfield_loader import load_ephemeris
from radioskysims.infrastructure.external.solar_flux_downloader import (
    SolarFluxDownloader,
)

# ── Define timerange ──
tr_solar = TimeRange.from_iso("2026-07-25T16:00:00", "2026-07-26T16:00:00", resolution=600)
print("== LocalSource ==")
print("  timerange:", tr_solar.min, "..", tr_solar.max)

# ── Define observador (BINGO) ──
obs_solar = Observer(lat=-7.25, lon=-35.86, height=550.0, timeframe=tr_solar)
print("  observer:", obs_solar.lat.to_value("deg"), obs_solar.lon.to_value("deg"))

# ── Uso do repositório (cache local + fallback de download) ──
service = LocalSourcesService(
    solar_flux_downloader=SolarFluxDownloader(),
    solar_flux_repository=SolarFluxRepository(
        engine=None,
        cache_dir=str(PROJECT_ROOT / "data/aux/signals/solar_flux"),
    ),
    local_source_track_provider=SkyfieldLocalSourceTrackProvider(),
)
print("  LocalSourcesService pronto (cache-first)")

# ── Obtém fluxo solar no intervalo ──
sf = service.get_solar_flux(
    center_utc=datetime(2026, 7, 26, 4, 0, tzinfo=UTC),
    window_hours=24,
)
print("  Fluxo solar:", sf.start_utc, "->", sf.end_utc)
print("  freqs (MHz):", sf.dataset["frequency_mhz"].values.tolist())
print("  dims:", dict(sf.dataset.sizes))

# ── Gráfico de fluxo solar freq × tempo com colorbar ──
flux_da = sf.dataset["flux_jy"].compute()

# Downsample temporal para plot leve (39k pontos → ~400)
step = max(1, flux_da.sizes["time"] // 400)
flux_sub = flux_da.isel(time=slice(None, None, step))

fig, ax = plt.subplots(figsize=(12, 5))
im = ax.pcolormesh(
    flux_sub.time.values,
    flux_sub.frequency_mhz.values,
    flux_sub.T.values,
    shading="auto",
    cmap="inferno",
)
ax.set_xlabel("Tempo (UTC)")
ax.set_ylabel("Frequência (MHz)")
ax.set_title("Fluxo solar — tempo × frequência")
plt.colorbar(im, label="Fluxo (Jy)")
plt.tight_layout()
plt.show()

# ── Skychart Mollweide com radiofontes e o Sol ──
# Posições do Sol no intervalo (RA/Dec via skyfield)
eph = load_ephemeris()
times = tr_solar.to_skyfield()
obs_loc, _ = obs_solar.to_skyfield()
app = (eph["earth"] + obs_loc).at(times).observe(eph["sun"]).apparent()
sun_ra = app.radec()[0].radians
sun_dec = app.radec()[1].radians

fig = plt.figure(figsize=(12, 6))
ax = fig.add_subplot(111, projection="mollweide")

# Radiofontes (se disponíveis da seção anterior; senão recria canônicas)
try:
    rs = display_sources
except NameError:
    from radioskysims.domain.radiosources import RadioSource, SpectralModel

    rs = [
        RadioSource(
            id="CygA",
            ra_deg=299.868,
            dec_deg=40.734,
            spectral_model=SpectralModel(nu0_mhz=1400.0, s0_jy=1570.0, alpha=0.75),
        ),
        RadioSource(
            id="CasA",
            ra_deg=350.867,
            dec_deg=58.815,
            spectral_model=SpectralModel(nu0_mhz=1400.0, s0_jy=2200.0, alpha=0.77),
        ),
        RadioSource(
            id="VirA",
            ra_deg=187.706,
            dec_deg=12.391,
            spectral_model=SpectralModel(nu0_mhz=1400.0, s0_jy=370.0, alpha=0.83),
        ),
    ]
for s in rs:
    ra_mw = ((np.deg2rad(s.ra_deg) + np.pi) % (2 * np.pi)) - np.pi
    ax.scatter(ra_mw, np.deg2rad(s.dec_deg), s=30, c="tab:blue", edgecolors="k")

# Sol: primeiro e último bem marcados, intermediários com alpha 0.2
ra_mw = ((sun_ra + np.pi) % (2 * np.pi)) - np.pi
n = len(sun_ra)
ax.plot(ra_mw, sun_dec, "-", color="orange", linewidth=0.8, alpha=0.6)
ax.scatter(ra_mw[0], sun_dec[0], s=140, c="gold", edgecolors="k", label="Sol (início)")
ax.scatter(ra_mw[-1], sun_dec[-1], s=140, c="darkorange", edgecolors="k", label="Sol (fim)")
mid = slice(1, -1)
ax.scatter(ra_mw[mid], sun_dec[mid], s=30, c="orange", alpha=0.2)

ax.grid(True, alpha=0.3)
ax.legend(loc="lower right")
ax.set_title("Skychart Mollweide — radiofontes + posições do Sol no intervalo")
plt.tight_layout()
plt.show()
```

- projetar tracks na superfície terreste (above method skyfield)
- construir heatmap com densidade de satélites no céu (repetir um para cada constelação GPS, GALILEO, BEIDOU, GLONASS)

```python
"""Satélites: repositório, GALILEO, tracks, projeção e heatmap de densidade."""

from __future__ import annotations

from datetime import datetime

import matplotlib.pyplot as plt
import numpy as np

from radioskysims.domain.common import Observer, TimeRange
from radioskysims.domain.geometry import SkyfieldGeometryProvider
from radioskysims.domain.tracks.factory import TrackFactory
from radioskysims.infrastructure.db.engine import Database
from radioskysims.infrastructure.db.satellite.satelliterepository import (
    SatelliteRepository,
)

tr_sat = TimeRange.from_iso("2026-07-25T00:00:00", "2026-07-25T02:00:00", resolution=600)
epoch = datetime(2026, 7, 25)  # naive, como esperado pelo repositório
db = Database()

# ── Uso do repositório ──
repo = SatelliteRepository(
    engine=db,
    tr=tr_sat,
    constellations=["GPS", "GALILEO", "BEIDOU", "GLONASS"],
)
print("== Satélites ==")
print("  NORAD IDs no catálogo:", len(repo.norad_ids))

# ── Seleciona uma constelação (Galileo) ──
gal_sats = repo.query_constellation(["GALILEO"], epoch=epoch)
print(f"  GALILEO: {len(gal_sats)} satélites")
if gal_sats:
    print("  exemplo:", gal_sats[0].norad_id, gal_sats[0].name)

# ── Calcula tracks ──
sat_blocks = {float(tr_sat.start_jd): {"timerange": tr_sat, "satellites": gal_sats}}
tf = TrackFactory()
tracks = tf.build(sat_blocks)
print("  tracks dims:", dict(tracks.dataset.sizes))

obs_sat = Observer(lat=-7.25, lon=-35.86, height=550.0, timeframe=tr_sat)
geo = SkyfieldGeometryProvider(tf.skyfield.ts).compute(tracks.dataset, obs_sat)
print("  geo vars:", list(geo.data_vars))

# ── Projeta tracks na superfície celeste (above horizon: el > 0) ──
fig = plt.figure(figsize=(12, 6))
ax = fig.add_subplot(111, projection="polar")
az = geo["az"].values
el = geo["el"].values
above = el > 0
theta_polar = (np.pi / 2 - az) % (2 * np.pi)  # 0=N, sentido horário
r_plot = np.pi / 2 - el  # 0=zenit, pi/2=horizonte
for i in range(geo.sizes["satellite"]):
    m = above[i]
    if not m.any():
        continue
    ax.plot(theta_polar[i][m], r_plot[i][m], "-", alpha=0.5, linewidth=0.7)
ax.set_ylim(0, np.pi / 2)
ax.set_yticks([0, np.pi / 6, np.pi / 3, np.pi / 2])
ax.set_yticklabels(["90°", "60°", "30°", "0°"])
ax.set_title("GALILEO — projeção celeste (el > 0)")
plt.tight_layout()
plt.show()

# ── Heatmap densidade de satélites no céu (por constelação) ──
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
for ax, const in zip(axes.flat, ["GPS", "GALILEO", "BEIDOU", "GLONASS"]):
    c_sats = repo.query_constellation([const], epoch=epoch)
    if not c_sats:
        ax.set_title(f"{const} — sem dados")
        ax.axis("off")
        continue
    c_blocks = {float(tr_sat.start_jd): {"timerange": tr_sat, "satellites": c_sats}}
    c_tracks = tf.build(c_blocks)
    c_geo = SkyfieldGeometryProvider(tf.skyfield.ts).compute(c_tracks.dataset, obs_sat)
    c_az = np.degrees(c_geo["az"].values)
    c_el = np.degrees(c_geo["el"].values)
    m = c_el > 0
    h = ax.hist2d(
        c_az[m],
        c_el[m],
        bins=(36, 18),
        range=[[0, 360], [0, 90]],
        cmap="inferno",
    )
    ax.set_xlabel("Azimute (deg)")
    ax.set_ylabel("Elevação (deg)")
    ax.set_title(f"{const} — densidade de satélites no céu")
    plt.colorbar(h[3], ax=ax, label="contagens")
plt.suptitle("Densidade de satélites no céu (2h, resolução 10 min)")
plt.tight_layout()
plt.show()
```

- Construir sinal do satélite e objeto xarray, mostrando gráfico de PSD e waterfall para um satelite de exemplo.

```python
"""Sinal do satélite: PSD e waterfall para um satélite de exemplo."""

from __future__ import annotations

from unittest.mock import MagicMock

import matplotlib.pyplot as plt
import numpy as np

from radioskysims.domain.common import FrequencyRange
from radioskysims.domain.signal.factory import SignalFactory

fr_gnss = FrequencyRange(min_mhz=980.0, max_mhz=1360.0, resolution_mhz=2.0)
sf = SignalFactory(frequency_range=fr_gnss, signal_repo=MagicMock())
grid_mhz = sf.frequency_grid_hz() * 1e-6

# PSD de um sinal GNSS (GPS L1: BPSK(1))
psd = sf._simulate_psd(
    modulation="BPSK(1)",
    fc_hz=1575.42e6,
    prn_len=2048,
    fs_factor=8,
).compute()

print("== Sinal GNSS (GPS L1, BPSK(1)) ==")
print("  grid:", grid_mhz.min(), "..", grid_mhz.max(), "MHz,", len(grid_mhz), "pontos")

# ── Gráfico de PSD ──
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(grid_mhz, psd, linewidth=0.8)
ax.set_xlabel("Frequência (MHz)")
ax.set_ylabel("PSD (dB/Hz)")
ax.set_title("PSD BPSK(1) — GPS L1")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ── Waterfall sintético (PSD repetida no tempo) ──
n_time = 30
waterfall = np.tile(psd, (n_time, 1))
fig, ax = plt.subplots(figsize=(10, 5))
im = ax.imshow(
    waterfall,
    aspect="auto",
    origin="lower",
    extent=[grid_mhz.min(), grid_mhz.max(), 0, n_time],
    cmap="inferno",
)
ax.set_xlabel("Frequência (MHz)")
ax.set_ylabel("Amostra de tempo")
ax.set_title("Waterfall — sinal do satélite (sintético)")
plt.colorbar(im, label="PSD (dB/Hz)")
plt.tight_layout()
plt.show()

# ── Objeto xarray (padrão do pipeline) ──
import xarray as xr

signals_ds = xr.Dataset(
    {"PSD": (("freq",), psd)},
    coords={"freq": grid_mhz},
)
print("Objeto xarray:")
print(signals_ds)
```

## Maps

- Constroi mapa sincroton pysm3 para frequency_range dividido em 10 intervalos
- Visualiza primeiro e ultimo em unica linha
- Downgrade
- Visualiza original e downgrades
- Calculo Cls original e downgraded e visualiza primeiro e ultimo
- Aplica marcara para declinação entre -30 e 30 graus
- faz rotações levando em conta apontamento do radiotelescopio e o timerange com benchmark

```python
"""Maps: mapa sincrotron pysm3, downgrade, Cls, máscara e rotações."""

from __future__ import annotations

import time

import healpy as hp
import matplotlib.pyplot as plt
import numpy as np

from radioskysims.application.skymap.dto import GenerateSkyMapInputDTO
from radioskysims.application.skymap.generate import GenerateSkyMapUseCase
from radioskysims.domain.skymap import NSide, SkyMapAnalysis
from radioskysims.domain.skymap.mask import SkyMask
from radioskysims.domain.skymap.operations import SkyMapOperations
from radioskysims.infrastructure.skymap.providers.pysm import (
    PySMSkyMapGeneratorAdapter,
)

NSIDE = 16
NSIDE_DOWN = 8
FREQS_GHZ = np.linspace(0.98, 1.36, 10)  # 980–1360 MHz em 10 intervalos

adapter = PySMSkyMapGeneratorAdapter(unit="uK_CMB", field="I")
uc = GenerateSkyMapUseCase(generator=adapter)

# ── Constrói mapa sincrotron pysm3 para cada intervalo ──
print("== Maps: sincrotron pysm3 ==")
maps = []
for i, f in enumerate(FREQS_GHZ):
    out = uc.execute(
        GenerateSkyMapInputDTO(
            frequency_ghz=float(f),
            nside=NSIDE,
            components=("synchrotron",),
        )
    )
    maps.append(out.map_data)
    print(f"  freq {f:.3f} GHz: shape={out.map_data.shape}, mean={out.map_data.mean():.2f}")

# ── Visualiza primeiro e último em uma linha ──
fig = plt.figure(figsize=(16, 6))
hp.mollview(
    maps[0],
    title=f"Synchrotron {FREQS_GHZ[0]:.2f} GHz",
    sub=(1, 2, 1),
    unit="uK_CMB",
    cmap="inferno",
)
hp.mollview(
    maps[-1],
    title=f"Synchrotron {FREQS_GHZ[-1]:.2f} GHz",
    sub=(1, 2, 2),
    unit="uK_CMB",
    cmap="inferno",
)
plt.show()

# ── Downgrade ──
downgraded = [SkyMapOperations.downgrade(m, NSide(NSIDE_DOWN)) for m in maps]
print("Downgrade:", maps[0].size, "->", downgraded[0].size, "pixels")

# ── Visualiza original e downgrades ──
fig = plt.figure(figsize=(16, 6))
hp.mollview(
    maps[0],
    title=f"Original nside={NSIDE} {FREQS_GHZ[0]:.2f} GHz",
    sub=(1, 2, 1),
    unit="uK_CMB",
    cmap="inferno",
)
hp.mollview(
    downgraded[0],
    title=f"Downgraded nside={NSIDE_DOWN} {FREQS_GHZ[0]:.2f} GHz",
    sub=(1, 2, 2),
    unit="uK_CMB",
    cmap="inferno",
)
plt.show()

# ── Cálculo de Cls original e downgraded, visualiza primeiro e último ──
cl_orig = hp.anafast(maps[0], lmax=3 * NSIDE - 1)
cl_down = hp.anafast(downgraded[0], lmax=3 * NSIDE_DOWN - 1)
ell_orig = np.arange(len(cl_orig))
ell_down = np.arange(len(cl_down))

fig, ax = plt.subplots(figsize=(10, 5))
ax.loglog(ell_orig, cl_orig, "o-", label=f"original nside={NSIDE}")
ax.loglog(ell_down, cl_down, "s-", label=f"downgraded nside={NSIDE_DOWN}")
ax.set_xlabel("ℓ")
ax.set_ylabel("Cℓ")
ax.set_title("Espectro angular (Cls)")
ax.legend()
ax.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()

# ── Aplica máscara para declinação entre -30° e 30° ──
theta, _ = hp.pix2ang(NSIDE, np.arange(maps[0].size))
dec_deg = 90.0 - np.degrees(theta)
mask = ((dec_deg >= -30) & (dec_deg <= 30)).astype(float)
masked = SkyMask.apply(maps[0], mask)
print("Máscara |dec|≤30°: fração de pixels preservados =", mask.mean())

fig = plt.figure(figsize=(10, 5))
hp.mollview(masked, title="Máscara declinação −30°..30°", unit="uK_CMB", cmap="inferno")
plt.show()

# ── Rotações levando em conta o apontamento do radiotelescópio + benchmark ──
from radioskysims.domain.common import TimeRange
from radioskysims.domain.simulation.pointing import build_pointing_track
from radioskysims.infrastructure.config.config_loader import (
    build_radiotelescope_from_settings,
)

rt_map = build_radiotelescope_from_settings("bingo")
tr_map = TimeRange.from_iso("2026-07-25T00:00:00", "2026-07-25T06:00:00", resolution=600)
obs_map = rt_map.observer
pt = build_pointing_track(rt_map.antennas[0], tr_map, obs_map)
ra0 = float(np.degrees(np.median(pt["ra"].values)))
dec0 = float(np.degrees(np.median(pt["dec"].values)))
rot = (np.radians(ra0), np.radians(90 - dec0), 0.0)

t0 = time.perf_counter()
rotator = hp.Rotator(rot=rot)
rotated = rotator.rotate_map_pixel(maps[0])
dt = time.perf_counter() - t0
print(f"Rotação para pointing (RA={ra0:.1f}°, Dec={dec0:.1f}°): {dt*1000:.1f} ms")

fig = plt.figure(figsize=(16, 6))
hp.mollview(maps[0], title="Original", sub=(1, 2, 1), unit="uK_CMB", cmap="inferno")
hp.mollview(rotated, title="Rotacionado (pointing BINGO)", sub=(1, 2, 2), unit="uK_CMB", cmap="inferno")
plt.show()
```

```python
"""TOD: criação da simulação, run com benchmark, propriedades, eventos e visualização."""

from __future__ import annotations

import time

from radioskysims.application.services.sim_orchestrator import SimOrchestrator
from radioskysims.domain.common import TimeRange
from radioskysims.visualization import time_series_plot, waterfall_plot

# ── Criação da simulação ──
orch = SimOrchestrator.from_settings("bingo")
print("== TOD ==")
print("  telescópio:", orch.context.telescope_name)
print("  repositório de simulação:", orch.simulation_repository is not None)

tr_tod = TimeRange.from_iso("2026-07-25T00:00:00", "2026-07-25T01:00:00", resolution=120)

# ── Roda simulação com benchmark ──
t0 = time.perf_counter()
comp = orch.run(
    constellations=["GALILEO"],
    time_range=tr_tod,
    use_processor=True,
)
dt = time.perf_counter() - t0
print(f"  run: {dt:.1f}s | n_satellites={comp.config.get('n_satellites')}")

# ── Mostra propriedades dos objetos ──
ds = comp.simulation_result.dataset
print("  dims:", dict(ds.sizes))
print("  vars:", list(ds.data_vars))
print("  time_range:", comp.simulation_result.time_range.min, "..", comp.simulation_result.time_range.max)
print("  freq_range (MHz):", comp.simulation_result.freq_range.min, "..", comp.simulation_result.freq_range.max)

# ── Mostra eventos ──
ev = comp.events
print("  eventos — objetos no feixe:", ev.objects_in_beam)
print("  T_antenna média (K):", ev.t_antenna_mean)
print("  duração (s):", ev.duration_s)

# ── Visualização ──
ds_computed = comp.compute()
waterfall_plot(
    ds_computed,
    source_type="satellite",
    title="Waterfall TOD — satélites (GALILEO)",
    use_datashader=False,
)
```
