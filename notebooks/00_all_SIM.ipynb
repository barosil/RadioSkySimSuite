## 3. Banco de Dados

Conecta ao PostgreSQL, diagnostica a conexão e consulta as tabelas principais.


```python
# ── 3a. Conexão com diagnóstico ──
from radioskysims.infrastructure.db.engine import Database
from sqlalchemy import text

db = Database()
print(f"📦 Database engine: {db.engine}")

# Diagnóstico: versão do PostgreSQL
try:
    with db.session() as session:
        result = session.exec(text("SELECT version()"))
        pg_version = result.one()
        print(f"🐘 PostgreSQL: {pg_version[0]}")
except Exception as e:
    print(f"❌ Erro de conexão: {e}")
```

```python
# ── 3b. Satélites — resumo por constelação ──
from sqlmodel import select, func
from radioskysims.infrastructure.db.satellite.models import SatID

with db.session() as session:
    # Total de satélites
    total = session.exec(select(func.count(SatID.norad_id))).one()
    print(f"🛰️  Total de satélites no catálogo: {total}")

    # Contagem por constelação
    const_q = session.exec(
        select(SatID.constellation, func.count(SatID.norad_id))
        .where(SatID.constellation.isnot(None))
        .group_by(SatID.constellation)
        .order_by(func.count(SatID.norad_id).desc())
    )
    print("\n📊 Satélites por constelação:")
    for const, cnt in const_q:
        print(f"   {const:20s}: {cnt}")
```

```python
# ── 3c. Elementos orbitais — resumo ──
from radioskysims.infrastructure.db.satellite.models import OrbitalElement

with db.session() as session:
    total_oe = session.exec(select(func.count(OrbitalElement.id))).one()
    latest_epoch = session.exec(select(func.max(OrbitalElement.epoch))).one()
    print(f"📐 Elementos orbitais no banco: {total_oe}")
    print(f"   Época mais recente: {latest_epoch}")

    # Top 5 satélites com elementos mais recentes
    recent = session.exec(
        select(OrbitalElement.norad_id, OrbitalElement.object_name, OrbitalElement.epoch)
        .where(OrbitalElement.epoch.isnot(None))
        .order_by(OrbitalElement.epoch.desc())
        .limit(5)
    )
    print("\n   Top 5 elementos mais recentes:")
    for norad, name, ep in recent:
        print(f"      NORAD {norad:6d} | {name or 'N/A':20s} | {ep}")
```

```python
# ── 3d. Rádio-fontes (NVSS) — resumo ──
from radioskysims.infrastructure.db.radiosources_db_model import RadioSourceDB

with db.session() as session:
    total_rs = session.exec(select(func.count(RadioSourceDB.id))).one()
    print(f"📡 Rádio-fontes no catálogo: {total_rs}")

    if total_rs > 0:
        # Estatísticas de fluxo
        stats = session.exec(
            select(
                func.min(RadioSourceDB.s0_jy),
                func.max(RadioSourceDB.s0_jy),
                func.avg(RadioSourceDB.s0_jy),
            )
        ).one()
        print(f"   Fluxo S₁.₄: min={stats[0]:.4f}  max={stats[1]:.4f}  média={stats[2]:.4f} Jy")

        # Distribuição de índice espectral
        alpha_stats = session.exec(
            select(
                func.min(RadioSourceDB.alpha),
                func.max(RadioSourceDB.alpha),
                func.avg(RadioSourceDB.alpha),
            )
        ).one()
        print(
            f"   Índice α: min={alpha_stats[0]:.3f}  max={alpha_stats[1]:.3f}  média={alpha_stats[2]:.3f}"
        )
    else:
        print("   ⚠️  Nenhuma rádio-fonte persistida. Execute a ingestão primeiro.")
```

```python
# ── 3e. Dados de fluxo solar — resumo ──
from radioskysims.infrastructure.db.solar_flux_db_model import SolarFluxDB

with db.session() as session:
    total_sf = session.exec(select(func.count(SolarFluxDB.id))).one()
    print(f"☀️  Registros de fluxo solar em cache: {total_sf}")

    if total_sf > 0:
        ranges = session.exec(
            select(
                func.min(SolarFluxDB.start_utc),
                func.max(SolarFluxDB.end_utc),
            )
        ).one()
        print(f"   Cobertura: {ranges[0]} a {ranges[1]}")

        # Lista arquivos
        files = session.exec(
            select(SolarFluxDB.filename, SolarFluxDB.start_utc, SolarFluxDB.end_utc)
            .order_by(SolarFluxDB.start_utc.desc())
            .limit(5)
        )
        print("   Últimos 5 arquivos:")
        for fname, s, e in files:
            print(f"      {fname}")
            print(f"         {s} → {e}")
    else:
        print("   ⚠️  Nenhum dado de fluxo solar em cache.")
```

## 7b. Pointing Track — RA/Dec Temporal via Skyfield

Demonstra a nova função `compute_pointing_track()` que converte um pointing alt-az fixo em RA/Dec aparente ao longo do tempo usando Skyfield. Com a rotação da Terra, um pointing fixo em alt-az varre diferentes RA/Dec — essencial para comparar com fontes de RA/Dec fixo (rádio-fontes).



```python
from radioskysims.domain.geometry import compute_pointing_track
from radioskysims.domain.common import Observer, TimeRange
import numpy as np
import matplotlib.pyplot as plt

# ── Observador BINGO ──
obs = Observer(
    lat=-7.25,
    lon=-35.86,
    height=550.0,
    timeframe=TimeRange.from_iso("2026-07-26T18:00:00", "2026-07-26T22:00:00", resolution=300),
)

# ── Grade temporal (4 horas, resolução 5 min) ──
time_grid = np.arange(
    np.datetime64("2026-07-28T18:00:00"),
    np.datetime64("2026-07-28T22:00:00"),
    np.timedelta64(300, "s"),
    dtype="datetime64[ns]",
)

# ── Pointing fixo alt-az → RA/Dec temporal ──
pt = compute_pointing_track(obs, time_grid, alt_deg=67.0, az_deg=0.0)
print(f"Pointing track: {pt.dims}, {len(pt.time)} timesteps")
print(f"  RA range:  {float(pt.ra.min().values):.2f}° – {float(pt.ra.max().values):.2f}°")
print(f"  Dec range: {float(pt.dec.min().values):.2f}° – {float(pt.dec.max().values):.2f}°")
print(f"  Variação RA (4h): {float(pt.ra.max().values - pt.ra.min().values):.2f}° (esperado ~60°)")

# ── Plot: RA/Dec × tempo ──
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

t_vals = (pt.time.values - pt.time.values[0]).astype("timedelta64[m]").astype(int)

ax1.plot(t_vals, pt.ra.values, "b-", linewidth=2)
ax1.set_ylabel("RA (graus)")
ax1.set_title("Pointing Track — RA/Dec Temporal (alt=67°, az=0°)")
ax1.grid(True, alpha=0.3)

ax2.plot(t_vals, pt.dec.values, "r-", linewidth=2)
ax2.set_xlabel("Tempo (minutos desde 18:00 UTC)")
ax2.set_ylabel("Dec (graus)")
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 7c. Theta Boresight com Pointing Temporal

A função `compute_theta_boresight()` agora aceita `pointing_track_ra`/`pointing_track_dec` (DataArrays temporais).
Isso permite calcular a separação angular ponto-a-ponto entre o pointing (que varia no tempo) e cada fonte.



```python
from radioskysims.domain.signal_source import compute_theta_boresight
import xarray as xr

# ── Cria tracks mock de uma rádio-fonte (RA/Dec fixos) ──
n = len(time_grid)
source_tracks = xr.Dataset(
    data_vars={
        "ra": ("time", np.deg2rad(np.full(n, 180.0))),  # RA fixa = 180°
        "dec": ("time", np.deg2rad(np.full(n, 0.0))),  # Dec fixa = 0°
        "theta": ("time", np.deg2rad(np.full(n, 15.0))),
        "az": ("time", np.deg2rad(np.full(n, 90.0))),
        "el": ("time", np.deg2rad(np.full(n, 75.0))),
        "range": ("time", np.full(n, 1e20, dtype=float)),
    },
    coords={"time": time_grid},
)

# ── Theta boresight com pointing temporal ──
gamma = compute_theta_boresight(
    source_tracks,
    pointing_track_ra=pt["ra"],
    pointing_track_dec=pt["dec"],
)

print(
    f"Theta boresight (temporal): {gamma.shape}, range = {float(gamma.min().values):.2f}–{float(gamma.max().values):.2f} rad"
)

# ── Comparação com pointing fixo ──
gamma_fixed = compute_theta_boresight(
    source_tracks,
    pointing_ra_deg=float(pt["ra"].mean().values),
    pointing_dec_deg=float(pt["dec"].mean().values),
)
print(
    f"Theta boresight (fixo):     range = {float(gamma_fixed.min().values):.2f}–{float(gamma_fixed.max().values):.2f} rad"
)

# ── Plot: diferença entre temporal e fixo ──
t_min = (time_grid - time_grid[0]).astype("timedelta64[m]").astype(int)
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(t_min, np.rad2deg(gamma.values), "b-", linewidth=2, label="Pointing temporal")
ax.plot(t_min, np.rad2deg(gamma_fixed.values), "r--", linewidth=2, label="Pointing fixo (médio)")
ax.set_xlabel("Tempo (min desde 18:00 UTC)")
ax.set_ylabel("Theta boresight (graus)")
ax.set_title("Separação Angular — Pointing Temporal vs Fixo")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

diff_deg = np.rad2deg(np.abs(gamma - gamma_fixed).max().values)
print(f"Diferença máxima: {diff_deg:.3f}° — relevante para beams estreitos!")
```

### 7d. Agrupamento de Rádio-Fontes (RadioSourceGroupSignalSource)

`RadioSourceGroupSignalSource` agrupa rádio-fontes co-localizadas (mesmo RA/Dec), somando suas PSDs e mantendo um track único. Isso reduz a complexidade do grafo dask de O(N) para O(grupos).



```python
from radioskysims.domain.signal_source_group import RadioSourceGroupSignalSource
from radioskysims.domain.signal_source import RadioSourceSignalSource


# ── Cria tracks mock ──
class _MockTrack:
    def __init__(self):
        import xarray as xr

        self.dataset = xr.Dataset(
            {
                "ra": ("time", np.deg2rad([180.0, 180.0])),
                "dec": ("time", np.deg2rad([0.0, 0.0])),
                "theta": ("time", [0.1, 0.1]),
                "az": ("time", [0.1, 0.1]),
                "el": ("time", [0.1, 0.1]),
                "range": ("time", [1e20, 1e20]),
            },
            coords={
                "time": np.array(["2026-07-28T18:00", "2026-07-28T20:00"], dtype="datetime64[ns]")
            },
        )


# ── 5 fontes: 3 na mesma posição, 2 isoladas ──
srcs = [
    RadioSourceSignalSource("src_A", _MockTrack(), np.array([1.0]), np.array([1e9])),
    RadioSourceSignalSource("src_B", _MockTrack(), np.array([2.0]), np.array([1e9])),
    RadioSourceSignalSource("src_C", _MockTrack(), np.array([3.0]), np.array([1e9])),
]

# ── Agrupar ──
groups = RadioSourceGroupSignalSource.group_by_position(srcs)
print(f"Fontes originais: {len(srcs)}")
print(f"Grupos:           {len(groups)}")
print()

for g in groups:
    st = getattr(g, "source_type", "radiosource")
    name = g.name
    psd = g.psd()
    print(f"  {st:20s} | {name:30s} | PSD={float(psd.values[0]):.2e} W/Hz")

# ── 3 fontes agrupadas → 1 grupo com PSD soma (1+2+3 = 6e-26 W/Hz) ──
assert len(groups) == 1, f"Esperado 1 grupo, obtido {len(groups)}"
g = groups[0]
assert g.source_type == "radiosource_group"
assert abs(float(g.psd().values[0]) - 6.0e-26) < 1e-30
print("\n✅ Agrupamento OK — PSD soma 6e-26 = 1e-26 + 2e-26 + 3e-26")
```

## 8. Simulação TOD e Waterfall Diagram

Executa a simulação completa com satélites + Sol + rádio-fontes e gera:
- Waterfall (tempo × frequência × potência)
- Espectro médio
- Série temporal
- Comparação com ruído


```python
from radioskysims.domain.common import TimeRange
from radioskysims.application.services.sim_orchestrator import SimOrchestrator

# ── Intervalo de observação curto para demo ──
tr_tod = TimeRange.from_iso("2026-07-26T18:00:00", "2026-07-26T20:00:00", resolution=600)

# ── Orquestrador BINGO ──
orch = SimOrchestrator.from_settings("bingo")
print(f"🔭 Telescópio: {orch.context.telescope_name}")
print(f"   Lat: {orch.context.observer.lat}°  Lon: {orch.context.observer.lon}°")
print(f"   Range: {tr_tod}")

# ── Executa simulação multi-fonte ──
result = orch.run(
    constellations=["GPS", "GALILEO"],
    time_range=tr_tod,
    include_local_sources=True,  # Sol
    include_radio_sources=True,  # NVSS (se houver dados)
    use_processor=True,
)
ds = result.dataset
print(f"✅ Simulação concluída: {dict(ds.sizes)}")
print(f"   Variáveis: {list(ds.data_vars.keys())}")

# ── Tipos de fonte ──
src_types = list(ds.coords["source_type"].values)
sat_mask = [t == "satellite" for t in src_types]
sun_mask = [t == "localsource" for t in src_types]
radio_mask = [t == "radiosource" for t in src_types]
print(f"   {sum(sat_mask)} satélites + {sum(sun_mask)} local + {sum(radio_mask)} rádio")
```

```python
# ── Waterfall Diagram ──
# Hora local (UTC-3 para BINGO)
time_local = [t - pd.Timedelta(hours=3) for t in pd.to_datetime(ds.coords["time"].values)]

# Potência total (agrega source + polarization)
power_total = ds["power"].sum(dim=["polarization", "source", "pointing"])
p2d = power_total.compute()

# Downsample em frequência para visualização
step_f = max(1, p2d.sizes["frequency"] // 150)
p2d_ds = p2d[:, ::step_f]

# ── Waterfall com HoloViews ──
hv_wf = hv.Image(
    (p2d_ds.frequency / 1e6, time_local, np.asarray(p2d_ds.values, dtype=float) * 1e3),
    ["Freq (MHz)", "Tempo (UTC-3)"],
    "mW/Hz",
).opts(
    cmap="inferno",
    width=650,
    height=350,
    title="Waterfall TOD — Satélites + Sol + Rádio-fontes",
    colorbar=True,
    tools=["hover"],
)
hv_wf
```

```python
# ── Espectro médio + Série temporal ──
spec = ds["power"].mean(dim=["time", "polarization", "source", "pointing"]).compute()
hv_spec = hv.Curve((spec.frequency / 1e6, spec.values * 1e3), "Freq (MHz)", "mW/Hz").opts(
    width=400, title="Espectro médio"
)

ts = ds["power"].isel(frequency=0).sum(dim=["polarization", "source", "pointing"]).compute()
hv_ts = hv.Curve((time_local, ts.values * 1e3), "Tempo (UTC-3)", "mW/Hz").opts(
    width=400, title=f"TOD @ {ds.frequency.values[0] / 1e6:.0f} MHz"
)

hv_spec + hv_ts
```

```python
# ── T_antenna por tipo de fonte ──
T_total = ds["T_antenna"].compute()

T_sat = T_total.isel(source=sat_mask).sum(dim="source") if any(sat_mask) else None
T_sun = T_total.isel(source=sun_mask).sum(dim="source") if any(sun_mask) else None
T_radio = T_total.isel(source=radio_mask).sum(dim="source") if any(radio_mask) else None

print(f"📊 T_antenna médio:")
print(f"   Total:    {float(T_total.mean().values):.2e} K")
if T_sat is not None:
    print(f"   Satélites: {float(T_sat.mean().values):.2e} K")
if T_sun is not None:
    print(f"   Sol:       {float(T_sun.mean().values):.2e} K")
if T_radio is not None:
    print(f"   Rádio:     {float(T_radio.mean().values):.2e} K")

# ── Curvas separadas por tipo ──
curves = []
ts_dt = time_local

if T_sat is not None:
    ts_sat = T_sat.isel(frequency=0).sum(dim=["polarization", "pointing"]).values
    curves.append(
        hv.Curve((ts_dt, ts_sat * 1e3), "Tempo (UTC-3)", "mW/Hz").opts(
            color="blue", title="Satélites"
        )
    )
if T_sun is not None:
    ts_sun = T_sun.isel(frequency=0).sum(dim=["polarization", "pointing"]).values
    curves.append(
        hv.Curve((ts_dt, ts_sun * 1e3), "Tempo (UTC-3)", "mW/Hz").opts(color="orange", title="Sol")
    )
if T_radio is not None:
    ts_radio = T_radio.isel(frequency=0).sum(dim=["polarization", "pointing"]).values
    curves.append(
        hv.Curve((ts_dt, ts_radio * 1e3), "Tempo (UTC-3)", "mW/Hz").opts(
            color="red", title="Rádio-fontes"
        )
    )

if curves:
    layout = hv.Layout(curves).opts(shared_datasource=False).cols(2)
    layout
else:
    print("Nenhuma curva para exibir.")
```

```python
# ── Ruído 50K + comparação ──
from radioskysims.domain.noise import NoiseSimulator

power_clean = ds["power"].sum(dim=["polarization", "source", "pointing"])
noisy = NoiseSimulator(42).simulate(ds, tsys=50.0, bandwidth=380e6, time_constant=60)

c = noisy["power"].isel(frequency=0).compute()
hv_noisy = hv.Curve((time_local, c.values * 1e3), "Tempo (UTC-3)", "mW/Hz").opts(
    color="red", title="TOD + Ruído 50K"
)

# Waterfall com ruído
noise_wf = noisy["power"].compute()
step_f2 = max(1, noise_wf.sizes["frequency"] // 150)
nwf_ds = noise_wf[:, ::step_f2]
nwf_dt = time_local[:: max(1, len(time_local) // len(nwf_ds.time))][: len(nwf_ds.time)]

hv_wf_noise = hv.Image(
    (nwf_ds.frequency / 1e6, nwf_dt, np.asarray(nwf_ds.values, dtype=float) * 1e3),
    ["Freq (MHz)", "Tempo (UTC-3)"],
    "mW/Hz",
).opts(cmap="inferno", width=650, height=350, title="Waterfall + Ruído 50K", colorbar=True)

std = float((noisy["power"] - power_clean).std().values)
print(f"σ(noise) = {std:.2e} W/Hz")
print(f"SNR     = {float(power_clean.mean().values) / std:.1f}")

hv_wf_noise + hv_noisy
```
